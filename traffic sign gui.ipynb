```python
from tkinter import *
from tkinter import filedialog
from PIL import Image, ImageTk
from tensorflow.keras.models import load_model
import numpy as np
```

```python
model = load_model("trafficsignmodel.h5")
```

```python
classes = {
    1: 'Speed limit (20km/h)',
    2: 'Speed limit (30km/h)',
    3: 'Speed limit (50km/h)',
    4: 'Speed limit (60km/h)',
    5: 'Speed limit (70km/h)',
    6: 'Speed limit (80km/h)',
    7: 'End of speed limit (80km/h)',
    8: 'Speed limit (100km/h)',
    9: 'Speed limit (120km/h)',
    10: 'No passing',
    11: 'No passing veh over 3.5 tons',
    12: 'Right-of-way at intersection',
    13: 'Priority road',
    14: 'Yield',
    15: 'Stop',
    16: 'No vehicles',
    17: 'Veh > 3.5 tons prohibited',
    18: 'No entry',
    19: 'General caution',
    20: 'Dangerous curve left',
    21: 'Dangerous curve right',
    22: 'Double curve',
    23: 'Bumpy road',
    24: 'Slippery road',
    25: 'Road narrows on the right',
    26: 'Road work',
    27: 'Traffic signals',
    28: 'Pedestrians',
    29: 'Children crossing',
    30: 'Bicycles crossing',
    31: 'Beware of ice/snow',
    32: 'Wild animals crossing',
    33: 'End speed + passing limits',
    34: 'Turn right ahead',
    35: 'Turn left ahead',
    36: 'Ahead only',
    37: 'Go straight or right',
    38: 'Go straight or left',
    39: 'Keep right',
    40: 'Keep left',
    41: 'Roundabout mandatory',
    42: 'End of no passing',
    43: 'End no passing veh > 3.5 tons'
}
```

```python
main = Tk()
main.title("Traffic sign identification")
titleLbl = Label(main,text="Traffic sign identification",font=("Arial", 20, "bold"))
titleLbl.pack(pady=10,padx=5)
predLbl = Label(main)
imageLbl = Label(main)

def predict(path):
    image = Image.open(path)
    image = image.resize((30,30))
    image = np.array(image)
    image = image/255
    image = np.expand_dims(image,axis=0)

    pred = model.predict(image)
    predclassi = np.argmax(pred,axis=1)[0]+1
    predclass = classes[predclassi]
    predLbl.config(text=predclass)

predict(r"C:\Users\morga\.cache\kagglehub\datasets\meowmeowmeowmeowmeow\gtsrb-german-traffic-sign\versions\1\Test\00002.png")

def classify(path):
    classifyBtn = Button(main,text="Classify",command=lambda:predict(path))
    predLbl.pack()
    classifyBtn.pack(padx=5,pady=10)
    #classifyBtn.pack_forget()

def upload():
    predLbl.pack_forget()
    try:
        file = filedialog.askopenfilename()
        print(file)
        image = Image.open(file)
        #image.thumbnail(((main.winfo_width()/3),(main.winfo_height/3)))
        image = image.resize((100,100))
        visImage = ImageTk.PhotoImage(image)
        imageLbl.config(image=visImage)
        imageLbl.image = visImage
        imageLbl.pack(padx=5,pady=10)
        classify(file)
        
        
    except:
        print("ERROR could not find image")


    

uploadBtn = Button(main,text="Upload",command=upload)
uploadBtn.pack()
main.mainloop()
```
