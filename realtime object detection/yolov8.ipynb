```python
import numpy as np
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("yolov8n.pt")
```

```python
camera = cv2.VideoCapture(0)
while 1:
    success,frame = camera.read()
    if not success:
        break
    results = model(frame)
    anFrame = results[0].plot()
    cv2.imshow("Video",anFrame)
    if cv2.waitKey(1) == 27:
        break

camera.release()
cv2.destroyAllWindows()
```
