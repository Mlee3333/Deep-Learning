```python
import numpy as np
text = "A text is a sequence of characters or words that conveys information. It can refer to physical or digital written material (such as books or articles) or a short electronic message sent via mobile devices"
text = text.lower()
newText = ""
chars = "abcdefghijklmnopqrstuvwxyz "
for char in text:
    if char in chars:
        newText += char

text = newText.split()
words = sorted(list(set(text)))
print(words)
```

```python
WtoI = {w:i for i,w in enumerate(words)}
ItoW = {i:w for i,w in enumerate(words)}
textInd = [WtoI[word] for word in text]
```

```python
seqLen = 4
X = []
y = []
for pos in range(len(textInd)-seqLen):
    X.append(textInd[pos:pos+seqLen])
    y.append(textInd[pos+seqLen])

X = np.array(X)
y=np.array(y)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN

model = Sequential()
model.add(SimpleRNN(128,input_shape=(X.shape[1],len(words)),activation="relu"))
model.add(Dense(len(words),activation="softmax"))
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
Xencoded = np.zeros((X.shape[0],X.shape[1],len(words)))
for i,seq in enumerate(X):
    for p,word in enumerate(seq):
        Xencoded[i,p,word] = 1
        
y = np.expand_dims(y,axis=1)
```

```python
model.fit(Xencoded,y,epochs=50,batch_size=12)
```

```python
def predict(seed):
    seed = seed.lower().split()
    seed = [w for w in seed if w in WtoI]
    if len(seed)<seqLen:
        seed = seed.rjust(seqLen)
    elif len(seed)>seqLen:
        seed = seed[-seqLen:]

    seedIndex = [WtoI[w] for w in seed]
    Xseed = np.zeros((1,seqLen,len(words)))
    for pos,index in enumerate(seedIndex):
        Xseed[0,pos,index] = 1
    pred = model.predict(Xseed)
    pred = np.argmax(pred[0])
    pred = ItoW[pred]    
    print(pred)
```

```python
predict("a books or articles")
```
