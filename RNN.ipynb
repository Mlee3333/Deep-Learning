```python
import numpy as np
text = "good morning"
chars = sorted(list(set(text)))
print(chars)
```

```python
chartoind = {ch:i for i,ch in enumerate(chars)}
indtochar = {i:ch for i,ch in enumerate(chars)}
print(chartoind)
print(indtochar)
```

```python
textind = [chartoind[char] for char in text]
print(textind)
```

```python
sequLen = 4
X = []
y = []
for pos in range(len(textind)-sequLen):
    X.append(textind[pos:pos+sequLen])
    y.append(textind[pos+sequLen])

print(X)
X = np.array(X)
print(y)
y=np.array(y)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN

model = Sequential()
model.add(SimpleRNN(128,input_shape=(X.shape[1],len(chars)),activation="relu"))
model.add(Dense(len(chars),activation="softmax"))
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
Xencoded = np.zeros((X.shape[0],X.shape[1],len(chars)))
#print(Xencoded)
for i,sequence in enumerate(X):
    #print(sequence)
    for p,character in enumerate(sequence):
        #print(p)
        #print(character)
        Xencoded[i,p,character] = 1

print(Xencoded)
```

```python
y = np.expand_dims(y,axis=1)
print(y)
```

```python
model.fit(Xencoded,y,epochs=50,batch_size=12)
```

```python
def predict(seed):
    if len(seed)<sequLen:
        seed = seed.rjust(sequLen)
    elif len(seed)>sequLen:
        seed = seed[-sequLen:]

    seedIndex = [chartoind[char] for char in seed]
    Xseed = np.zeros((1,sequLen,len(chars)))
    for pos,index in enumerate(seedIndex):
        Xseed[0,pos,index] = 1
    pred = model.predict(Xseed)
    print(pred)
    pred = np.argmax(pred[0])
    print(pred)
    pred = indtochar[pred]    
    print(pred)
```
